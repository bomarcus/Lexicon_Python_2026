```python
import pandas as pd
import seaborn as sns

sns.set_theme()

movies_source = pd.read_csv("data/movies/movies.csv")
movies_test = movies_source.copy()
sns.set_style("ticks")
sns.relplot(
    data=movies_test,
    x="Production Budget",
    y="Worldwide Gross",
    col="MPAA Rating",
    col_wrap=3,
    hue="Creative Type",
    size="IMDB Rating",
)

sns.displot(movies_test, x="Production Budget", bins="auto")
movies_test.hist(figsize=(14, 10), bins=30)
```
