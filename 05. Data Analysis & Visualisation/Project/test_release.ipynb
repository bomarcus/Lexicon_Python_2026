```python
import calendar

import pandas as pd
import seaborn as sns
from pandas.tseries.api import guess_datetime_format

movies_source = pd.read_csv("data/movies/movies.csv")
movies = movies_source.copy()
```

```python
# check five first and five last rows
display(movies.head(), movies.tail())
```

month, day, year


```python
# visually check date format
display(movies["Release Date"].head(), (movies["Release Date"].tail()))
```

```python
# check for missing values
display(movies["Release Date"].isna().sum())
display(movies["Release Date"].isnull().sum())
```

```python
# testing the guess_datetime_format-function
guess_datetime_format("Jun 12 1998")
```

```python
# convert date
movies["Release Date"] = pd.to_datetime(
    movies["Release Date"], errors="coerce", format="%b %d %Y"
)
display(movies["Release Date"].head(10))
```

```python
# check 10 first and 10 last entries
display(movies["Release Date"].sort_values().dt.year.head(10))
display(movies["Release Date"].sort_values().dt.year.tail(10))
```

```python
# check if any values are missing after conversion
display(movies["Release Date"].isna().sum())
```

```python
# find first and last date
display(movies["Release Date"].min(), movies["Release Date"].max())
```

```python
sns.set_theme(style="whitegrid", palette="pastel")

ax = sns.countplot(x=movies["Release Date"].dt.month)
ax.set_title(
    f"Month of release over the years, {movies['Release Date'].dt.year.min()} - {movies['Release Date'].dt.year.max()}"
)
ax.set_xticks(range(12))
ax.set_xticklabels(calendar.month_abbr[1:13])
ax.set_xlabel("Month")
ax.set_ylabel("Number of releases")
```
