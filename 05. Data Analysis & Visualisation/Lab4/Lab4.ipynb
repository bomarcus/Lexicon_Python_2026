## 4.7

```python
members = members.drop_duplicates().reset_index(drop=True)
```

```python
print(members.shape)
```

```python
merged = loans.merge(
    members, on="member_id", how="left", validate="many_to_one", indicator=True
)
print(merged.shape)
print(merged["late_fee"].sum().round(1))
print((merged["_merge"] == "both").sum())
print((merged["_merge"] == "left_only").sum())
print((merged["_merge"] == "right_only").sum())
```

```python
no_member = merged[merged["_merge"] == "left_only"]
print(no_member["member_id"].nunique())
print(no_member["late_fee"].sum().round(1))

inner = loans.merge(members, on="member_id", how="inner")
print(inner.shape)
print(inner["late_fee"].sum().round(1))
```
