```python
# strip spaces
# make lowercase
# replace "children's" with "children"
# replace "sci fi"  and "science fiction" with "sci-fi"
# return distinct rows.
loans["genre"].str.strip().str.lower().replace(
    {"children's": "children", "sci fi": "sci-fi", "science fiction": "sci-fi"}
).unique()
```

```python
# strip spaces
# make lowercase
# replace "children's" with "children"
# replace "sci fi"  and "science fiction" with "sci-fi"
# add capitol first letter with title.
# save it back into the genre column.
loans["genre"] = (
    loans["genre"]
    .str.strip()
    .str.lower()
    .replace(
        {"children's": "children", "sci fi": "sci-fi", "science fiction": "sci-fi"}
    )
    .str.title()
)

# show genres and loans on each genre
loans["genre"].value_counts()
```

```python
loans = loans_raw.copy()

loans["genre"] = (
    loans["genre"]
    .str.strip()
    .str.lower()
    .replace(
        {"children's": "children", "sci fi": "sci-fi", "science fiction": "sci-fi"}
    )
    .str.title()
)

loans["returned"] = (
    loans["returned"]
    .str.strip()
    .str.lower()
    .map({"yes": True, "y": True, "no": False, "n": False})
)

loans["late_fee"] = (
    loans["late_fee"]
    .str.strip()
    .str.replace("kr", "")
    .str.replace(" ", "")
    .str.replace(",", ".")
    .astype(float)
)

loans["loan_date"] = pd.to_datetime(loans["loan_date"], errors="coerce")

loans = loans.drop_duplicates().reset_index(drop=True)

loans = loans.drop(columns="notes")

print(loans.shape)
loans.to_csv("data/loans_clean.csv", index=False)
```
