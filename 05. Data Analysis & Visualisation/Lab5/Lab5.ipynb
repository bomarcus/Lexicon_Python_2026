## 5.1

```python
import pandas as pd

rentals_raw = pd.read_csv("data/bike_rentals.csv")

rentals = rentals_raw.copy()
```

```python
rentals.shape
```

```python
print(rentals.dtypes)
print("-----")
print("price type: ", rentals["price"].dtype)
```

```python
empty = rentals.columns[rentals.isna().all()]
```

```python
print("columns where every value is missing:", len(empty))
```

```python
print("missing columns:", *empty, sep="\n")
```

```python
print("missing columns:", *empty, sep="\n")
```

```python
rentals = rentals.dropna(axis=1, how="all")

print(rentals.shape)
print("empty columns: ", rentals.isna().all().sum())
```
